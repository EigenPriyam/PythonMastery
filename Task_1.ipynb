### Print the given strings as per stated format.

**Given strings**: 
```
"Data" "Science" "Mentorship" "Program" 
"By" "CampusX"
```
**Output**: 
``` 
Data-Science-Mentorship-Program-started-By-CampusX
```

````python
s = ["Data" ,"Science" ,"Mentorship" ,"Program" ,"By" ,"CampusX"]
#Approach 1
print("-".join(s))

#Apporach 2
for i in s:
    print(i,end='-')
````

### Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

````python
input_1 = int(input("Please enter first number..."))
input_2 = int(input("Please enter second number..."))

c = input_1
input_1 = input_2
input_2 = c

print('Input 1' , input_1)
print('Input 2' , input_2)
````

### Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
print("Enter First Coordinate...")
x1 = float(input("Please enter first number..."))
y1 = float(input("Please enter second number..."))
print('x1' , ':' , x1 , "," , 'y1 : ', y1)


print("Enter Second coordinate...")
x2 = float(input("Please enter first number..."))
y2 = float(input("Please enter second number..."))
print('x2' , ':' , x2 , "," , 'y2 : ', y2)

print(f"Euclidean distance between two coordinates are {((x1-x2)**2 + (y1-y2)**2)**0.5}")
````
