```python
import traceback


class InsufficientFundsError(Exception):
    """Custom exception: raised when balance is too low."""


class ATM:
    def __init__(self, pin, balance=0):
        self.__pin = pin
        self.__balance = balance

    def check_pin(self, entered_pin):
        if entered_pin != self.__pin:
            raise ValueError("Incorrect PIN")

    def withdraw(self, pin, amount):
        self.check_pin(pin)
        if amount <= 0:
            raise ValueError(f"Invalid amount: {amount}")
        if amount > self.__balance:
            raise InsufficientFundsError(
                f"Balance {self.__balance} is less than requested {amount}"
            )
        self.__balance -= amount
        return self.__balance

    def deposit(self, pin, amount):
        self.check_pin(pin)
        self.__balance += int(amount)  # int("abc") -> ValueError, int(None) -> TypeError
        return self.__balance


atm = ATM(pin=1234, balance=1000)

# --------------------------------------------------------------
# 1. Basic try / except / else / finally
# --------------------------------------------------------------
print("=== 1. Structure ===")
try:
    print("Withdraw OK, new balance:", atm.withdraw(1234, 200))
except ValueError:
    print("Bad PIN or amount")
else:  # runs only if NO exception happened
    print("else: no error occurred")
finally:  # ALWAYS runs
    print("finally: card returned")

# --------------------------------------------------------------
# 2. `as e` -> e is the exception OBJECT.  str(e) vs e vs repr(e)
# --------------------------------------------------------------
print("\n=== 2. str(e) vs e ===")
try:
    atm.withdraw(1234, 5000)
except InsufficientFundsError as e:
    print("str(e)  :", str(e))        # just the message
    print("e       :", e)             # print() calls str() -> same as str(e)
    print("repr(e) :", repr(e))       # ClassName('message') -> best for logs/debugging
    print("type(e) :", type(e).__name__)
    print("e.args  :", e.args)        # tuple of arguments given to the exception
    print(f"f-string {{e}}  : {e}")
    print(f"f-string {{e!r}}: {e!r}")

# --------------------------------------------------------------
# 3. e.with_traceback(tb) and e.__traceback__
# --------------------------------------------------------------
print("\n=== 3. traceback ===")
try:
    atm.withdraw(9999, 100)  # wrong pin
except ValueError as e:
    tb = e.__traceback__  # the traceback object attached to e
    print("Failed at line:", tb.tb_lineno)   # line in THIS file where the try body failed
    print("Full traceback via traceback module:")
    traceback.print_tb(tb)

    # with_traceback(tb): sets e.__traceback__ = tb and RETURNS e.
    # Typical use: raise a NEW exception but keep the ORIGINAL traceback.
    try:
        raise RuntimeError("ATM service failed").with_traceback(tb)
    except RuntimeError as new_e:
        print("\nnew_e:", repr(new_e))
        print("Traceback still points to original failure:")
        traceback.print_tb(new_e.__traceback__)

# --------------------------------------------------------------
# 4. Multiple excepts, tuple of exceptions, generic Exception
# --------------------------------------------------------------
print("\n=== 4. Multiple excepts ===")
for amount in ["50", "abc", None]:
    try:
        print("Deposit ->", atm.deposit(1234, amount))
    except (ValueError, TypeError) as e:  # catch several types at once
        print(f"{type(e).__name__}: {e}")
    except Exception as e:  # catch-all; keep it LAST
        print("Unexpected:", repr(e))

# --------------------------------------------------------------
# 5. raise ... from e  (exception chaining)
# --------------------------------------------------------------
print("\n=== 5. Chaining ===")
try:
    try:
        atm.withdraw(1234, 10**6)
    except InsufficientFundsError as e:
        raise RuntimeError("Transaction declined") from e
except RuntimeError as e:
    print("outer :", e)
    print("cause :", repr(e.__cause__))  # the original exception
```

```python
class MyClass:
    counter = 10

    @staticmethod
    def static_method():
        print("Static method called. Counter =", MyClass.counter)
        MyClass.counter += 1  # modify class variable

    @classmethod
    def class_method(cls):
        print("Class method called. Counter =", cls.counter)
        cls.counter += 1  # modify class variable

obj = MyClass()
obj.static_method()
obj.class_method()

MyClass.counter
```
