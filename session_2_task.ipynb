### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
CTC = int(input('Please Enter Your CTC'))

salary_before_tax = CTC - (0.10*CTC) - (0.05*CTC) - (0.03*CTC)

final_salary = 0
if salary_before_tax < 500000:
    print("You are taxed at zero percentage .. Please enjoy your salary")
    final_salary = salary_before_tax
elif salary_before_tax > 500000 and salary_before_tax < 100000:
    print("You are taxed at 10 percentage slab ..")
    final_salary = salary_before_tax - (0.10*salary_before_tax)
elif salary_before_tax > 100000 and salary_before_tax < 200000:
    print("You are taxed at 20 percentage slab ..")
    final_salary = salary_before_tax - (0.20*salary_before_tax)
else:
    print("You are taxed at 30 percentage slab ..")
    final_salary = salary_before_tax - (0.30*salary_before_tax)
    
print("Your Final Salary is :" , final_salary)
````

### `Problem 4`: Write a menu-driven program -
1. cm to ft
2. km to miles
3. USD to INR
4. exit

````python
def cm_to_ft():
    cm = float(input("Enter length in centimeters: "))
    feet = cm / 30.48
    print(f"{cm} cm = {feet:.2f} ft")

def km_to_miles():
    km = float(input("Enter distance in kilometers: "))
    miles = km * 0.621371
    print(f"{km} km = {miles:.2f} miles")

def usd_to_inr():
    usd = float(input("Enter amount in USD: "))
    rate = 83.50  # example fixed exchange rate
    inr = usd * rate
    print(f"${usd} = ₹{inr:.2f}")

def main():
    while True:
        print("\n----- Menu -----")
        print("1. cm to ft")
        print("2. km to miles")
        print("3. USD to INR")
        print("4. exit")

        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            cm_to_ft()
        elif choice == '2':
            km_to_miles()
        elif choice == '3':
            usd_to_inr()
        elif choice == '4':
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 4.")


if __name__ == "__main__":
    main()
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

###`Problem 10`: Write a program, which will find all such numbers between 1000 and 3000 (both included) such that each digit of the number is an even number. The numbers obtained should be printed in a space-separated sequence on a single line.



````python
def function_check_even(num):
    is_even = True
    while num !=0:
        digit = num % 10
        if digit % 2 != 0:
            is_even = False
        num = num // 10
    return is_even

for i in range(1000 , 3001):
    if function_check_even(i) == True:
        print(i,sep=',')
````

###`Problem 11`: A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps.
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps.

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.**

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
`Output`:
```
2
```

````python
import math

pos = [0, 0]

while True:
    line = input().strip()
    if line == '!':
        break
    direction, steps = line.split()
    steps = int(steps)

    if direction == 'UP':
        pos[1] += steps
    elif direction == 'DOWN':
        pos[1] -= steps
    elif direction == 'LEFT':
        pos[0] -= steps
    elif direction == 'RIGHT':
        pos[0] += steps

distance = math.sqrt(pos[0]**2 + pos[1]**2)
print(round(distance))
````

###`Problem 12`:Write a program to print whether a given number is a prime number or not

````python
num = int(input("Enter a number: "))

if num < 2:
    print(f"{num} is not a prime number")
else:
    is_prime = True
    for i in range(2, num):          # check every number from 2 to num-1
        if num % i == 0:              # found a divisor other than 1 and itself
            is_prime = False
            break
    if is_prime:
        print(f"{num} is a prime number")
    else:
        print(f"{num} is not a prime number")
````

###`Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python
lower = int(input("Enter lower range: "))
upper = int(input("Enter upper range: "))

print(f"Armstrong numbers between {lower} and {upper}:")

for num in range(lower, upper + 1):
    digits = str(num)
    n = len(digits)
    total = sum(int(d) ** n for d in digits)
    if total == num:
        print(num)
````

###`Problem 14`:Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python
H = int(input("Enter hours (0-11 or 0-23): "))
M = int(input("Enter minutes (0-59): "))

H = H % 12  # normalize in case 24-hour format is given

# minute hand moves 6 degrees per minute
minute_angle = 6 * M

# hour hand moves 30 degrees per hour + 0.5 degrees per minute
hour_angle = 30 * H + 0.5 * M

angle = abs(hour_angle - minute_angle)

# the angle between hands is never more than 180; take the smaller one
angle = min(angle, 360 - angle)

print(int(angle))  # floor of the result
````
